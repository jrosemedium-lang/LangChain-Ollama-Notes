```python
from langchain_core.prompts import PromptTemplate
from langchain_ollama import ChatOllama


prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms"
)

model = ChatOllama(model="qwen2.5:1.5b")

chain = prompt | model

response = chain.invoke({"topic": "Java"})

#had to do response.content because 
#the response object is a LangChain Response object, 
#not a string.
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms",
)

model = ChatOllama(model="qwen2.5:1.5b", num_predict=5)

#With StrOutputParser, it returns the response text directly.
#StrOutputParser extracts text; 
#other parsers can produce structured outputs such as JSON.
parser = StrOutputParser()

#use pipe operator to chain prompt, model,& parser together.
chain = prompt | model | parser


#passing the RESULT OF PIPE to the parser
response = chain.invoke({"topic": "wkhkjh"})
print(response)
```
