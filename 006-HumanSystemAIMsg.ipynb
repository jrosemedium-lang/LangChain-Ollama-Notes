```python
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

messages = [
    # Instructions for the AI
    SystemMessage(
        content="You are a Java tutor. Explain concepts in simple language."
    ),

    # User's question
    HumanMessage(
        content="What is HashMap in Java?"
    ),

    # AI's previous response
    AIMessage(
        content="HashMap stores data in key-value pairs."
    ),

    # User's next question
    HumanMessage(
        content="Does it allow duplicate keys?"
    )
]
```

```python
from langchain_ollama import ChatOllama

chat = ChatOllama(model="qwen2.5:1.5b",  num_predict=20)

response = chat.invoke(messages)

print(response.content)
```
