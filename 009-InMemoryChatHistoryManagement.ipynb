```python

from langchain_core.chat_history import InMemoryChatMessageHistory

# Create chat history
history = InMemoryChatMessageHistory()

# Store messages
history.add_user_message("What is Java?")
history.add_ai_message("Java is an object-oriented programming language.")

history.add_user_message("What is Spring Boot?")
history.add_ai_message("Spring Boot simplifies Spring application development.")

# Retrieve all messages
for message in history.messages:
    print(message.type, ":", message.content)
```
