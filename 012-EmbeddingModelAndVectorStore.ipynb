```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

documents = [
    "Java is a programming language.",
    "Spring Boot is a framework built on Java.",
    "Python is widely used in machine learning.",
    "Hibernate is an ORM framework used with Java."
]

# documents = [
#     "Java",   
#     "Spring Boot",
#     "Python",
#     "Hibernate"
# ]

#Initialize the Embedding model
embedding_model = OllamaEmbeddings(model="nomic-embed-text")

#Create a vector store
#FAISS will convert each document into an embedding
#and stores all vectors along with documents
vector_store = FAISS.from_texts(documents, embedding_model)


# for i in range(vector_store.index.ntotal):
#    print(f"Document {i}: {vector_store.index.reconstruct(i)}")

print(vector_store.index.ntotal);

#Query from user to check semantically similar document
query = "To do Machine Learning, which programming language is best?"


results = vector_store.similarity_search(
    query,
    k=1
)

for document in results:
    print(document.page_content)

print("================================")

#default FAISS L2 distance configuration, a lower distance means a closer match
results = vector_store.similarity_search_with_score(
    query,
    k=3
) 

for document, score in results:
    print(document.page_content)
    print(score)


print("================================")
#Returns documents with normalized relevance scores, generally between 0 and 1. Higher means more relevant. This is different from the raw FAISS distance.
results = vector_store.similarity_search_with_relevance_scores(
    "Why did Obama used butter on tshirt?",
    k=1
)

for document, score in results:
    print(document.page_content)
    print(score)

# https://www.linkedin.com/pulse/langchain-retrieval-rutam-bhagat-d7kzf
print("================================")

results = vector_store.max_marginal_relevance_search(
    query,
    k=1,
    fetch_k=10,
    lambda_mult=0.5
)

for document in results:
    print(document.page_content)



print("================================")

#Here, you supply an already-generated numerical vector instead of query text. This means the embedding operation is not repeated inside the text-based search method.

query_vector = embedding_model.embed_query(query)

results = vector_store.similarity_search_by_vector(
    query_vector,
    k=2

)
```
