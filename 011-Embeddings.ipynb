```python
documents = [
    "Java is a programming language.",
    "Python is used for machine learning.",
    "Spring Boot is a Java framework."
]
```

```python
from langchain_ollama import OllamaEmbeddings
#OllamaEmbeddings connects LangChain to an embedding model.
embeddings = OllamaEmbeddings(model="nomic-embed-text")

#embed_query() converts your question into a numerical vector.
vector = embeddings.embed_query(
    "Which framework is used with Java?"
)

#The vector represents the semantic meaning of your question.
print(vector)

#No LLM answer generation happens here. We are only converting text into vectors.
```
