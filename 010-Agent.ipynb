```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_ollama import ChatOllama
```

```python
@tool
def get_day_for_letter(a:str) -> str:
    """Get the day of the week for a given letter."""
    print(f"Received letter: {a}")
    days = {
        'M': 'Monday',
        'T': 'Tuesday',
        'W': 'Wednesday',
        'R': 'Thursday',
        'F': 'Friday',
        'S': 'Saturday',
        'U': 'Sunday'
    }
    return days.get(a.upper(), "Invalid letter")
```

```python
llm = ChatOllama(
    model="qwen2.5:1.5b",
)

agent = create_agent(
    model=llm,
    tools=[get_day_for_letter],
    system_prompt="You are a helpful assistant."
)
```

```python
response = agent.invoke({
    "message":[
        {"role": "user", 
         "content": "What day of the week is represented by the letter 'M'? Check the tool and answer"}
    ]
})

print(response)
```
